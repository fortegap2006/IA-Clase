# Actividad de Clase: Analizando Agentes de IA con Hugging Face Spaces

## Objetivo

Explorar aplicaciones reales de Inteligencia Artificial en **Hugging Face Spaces** y analizarlas desde la perspectiva de los **agentes racionales**.

Al finalizar la actividad, los estudiantes deberán ser capaces de:

- Identificar los componentes **PEAS** de un agente.
- Clasificar las propiedades del entorno.
- Proponer qué tipo de programa de agente podría implementarse detrás del sistema.
- Justificar sus respuestas.

------------------------------------------------------------------------

## Instrucciones

1. Ingresen a **https://huggingface.co/spaces**.
2. Exploren diferentes Spaces.
3. Seleccionen uno que les parezca interesante.
4. Interactúen con el sistema durante algunos minutos.
5. Completen la siguiente ficha de análisis.

------------------------------------------------------------------------

# Ficha de análisis

## 1. Nombre del Space

**Nombre:** Z-Image-Turbo

**Enlace:** https://huggingface.co/spaces/mrfakename/Z-Image-Turbo

------------------------------------------------------------------------

## 2. ¿Qué hace el agente?

Describa en dos o tres líneas cuál es la función del sistema.

**Respuesta:** El agente genera imágenes a partir de una descripción del usuario. También permite especificar el ancho y el alto de la imagen.

------------------------------------------------------------------------

## 3. Análisis PEAS

| Elemento | Respuesta |
|---|---|
| **Performance** | El agente hace bien su trabajo cuando genera una imagen que se parezca lo más posible a la descripción del usuario. Mientras más fiel sea el resultado, mejor es su desempeño. |
| **Environment** | Una interfaz gráfica y los modelos de generación de imágenes que usa el sistema. |
| **Actuators** | Produce como salida una imagen generada. |
| **Sensors** | Recibe un prompt con la descripción de la imagen, y posiblemente parámetros como ancho y alto. |

------------------------------------------------------------------------

## 4. Clasificación del entorno

Complete la siguiente tabla y justifique brevemente cada respuesta.

| Propiedad | Clasificación | Justificación |
|---|---|---|
| Observabilidad | Parcialmente observable | No se puede ver directamente cómo el modelo interno procesa la información o qué representación exacta usa para generar la imagen. |
| Determinismo | No determinista | Con el mismo prompt, la imagen generada puede variar. |
| Episódico | No | La generación puede depender de parámetros o ajustes previos y no siempre se trata como una sola decisión aislada. |
| Estático | Sí | Una vez que se envía el prompt, el entorno no cambia continuamente durante la generación. |
| Discreto | Sí | Las acciones y entradas se manejan en pasos definidos, no como valores continuos. |
| Conocido | No | El comportamiento interno del modelo no es completamente conocido por el usuario. |

------------------------------------------------------------------------

## 5. ¿Qué tipo de programa de agente creen que es?

Seleccione la opción que consideren más adecuada y explique por qué.

- Agente de reflejo simple
- Agente basado en modelo
- Agente basado en objetivos
- Agente basado en utilidad
- Agente con aprendizaje

> **Importante:** No existe una única respuesta correcta. Lo importante es justificar la elección a partir del comportamiento observado.

**Respuesta:** Agente basado en objetivos. El sistema recibe una meta concreta: generar una imagen que cumpla con la descripción del usuario. Su comportamiento está orientado a alcanzar ese objetivo.
